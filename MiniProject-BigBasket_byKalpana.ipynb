## <u> <b>Step 8 : Find out the outliers in the dataset and fill them with mean value

```python

numerical_columns = ['sale_price', 'rating', 'market_price', 'Discount%']
plt.figure(figsize=(5, 15)) 
for i, column in enumerate(numerical_columns, 1):
 plt.subplot(len(numerical_columns), 1, i) 
 sns.boxplot(df[column], orient='h',color='#2B7A78')
 plt.title(f'Box Plot - {column}')
 plt.xlabel(column)
 plt.tight_layout() 
```

### a) Find out the outliers in the dataset

```python
def find_outliers(d,col):
 q1=d[col].quantile(0.25)
 q3=d[col].quantile(0.75)
 IQR=q3-q1
 lower_limit= q1-1.5*IQR
 upper_limit= q3+1.5*IQR
 return lower_limit,upper_limit

a= ['sale_price',	'market_price',	'Discount', 'rating']
for i in a:
 print(find_outliers(df,i))


```

### b) Replacing outliers with mean value

```python
df["sale_price"]=df["sale_price"].apply(lambda x: df["sale_price"].mean() if(x<-267 or x>694.12) else x)
df["sale_price"].head(10)
```

```python
df["market_price"]=df["market_price"].apply(lambda x: df["market_price"].mean() if(x<-323 or x>805) else x)
df["market_price"].tail(10)
```

```python
df["Discount"]=df["Discount"].apply(lambda x: df["Discount"].mean() if(x<-30 or x>50) else x)
df["Discount"].tail(10)
```

```python
df["rating"]=df["rating"].apply(lambda x: df["rating"].mean() if(x<3.91 or x>4.00) else x)
df["rating"].head(10)
```

## 9.1 Number of Products per Category
##### This bar plot shows the count of products within each category, providing insight into the distribution of products across various categories.


```python
counts = df['category'].value_counts()
counts
```

```python
counts_df = pd.DataFrame({'Category': counts.index, 'Counts': counts.values})
ax = sns.barplot(x='Category', y='Counts', data=counts_df, color='#2B7A78')
plt.xticks(rotation=75, ha='right')
ax.set_xlabel('Category')
ax.set_ylabel('No. of Products')
ax.set_title('No. of Products per Category')
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),ha='center', va='baseline', 
                fontsize=10, color='black', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
plt.show()
```

### <u> Key Takeaways</u>:
##### • Beauty & Hygiene dominates with the highest number of products with a volume of 7876 units.
##### • Gourmet & World Food and Kitchen, Garden & Pets are average selling products with 4690 and 3580 units, respectively.

##### • Beverages, Bakery, Cakes & Dairy  are the products with a moderate volume of approximately 900 units.
##### • Baby Care Fruits & Vegetables have significantly fewer products of around 600 units each.
##### • Eggs, Meat & Fish has the lowest product count, with 350 units.
